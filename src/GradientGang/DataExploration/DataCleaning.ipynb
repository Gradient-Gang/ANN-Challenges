# Data Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
DirectoryRaw = '../../../dataset/Pirate'
DirectoryCleaned = '../../../dataset/PirateCleaned'
```

## Import Data

```python
X_train = pd.read_csv(f'{DirectoryRaw}/pirate_pain_train.csv')
y_train = pd.read_csv(f'{DirectoryRaw}/pirate_pain_train_labels.csv')

X_test = pd.read_csv(f'{DirectoryRaw}/pirate_pain_test.csv')
```

## Display 5 timeseries

```python
Set = np.random.randint(0,100,size=10)
T = [ X_train[X_train["sample_index"] == i] for i in Set ]

print(f"Selected sample indices: {Set}")

fig, axes = plt.subplots(3, 4, figsize=(20, 10))
axes = axes.flatten()

for i, ti in enumerate(T):
    axes[0].plot(ti["time"], ti["pain_survey_1"])
    axes[1].plot(ti["time"], ti["pain_survey_2"])
    axes[2].plot(ti["time"], ti["pain_survey_3"])
    axes[3].plot(ti["time"], ti["pain_survey_4"])
    axes[4].plot(ti["time"], ti["n_legs"])
    axes[5].plot(ti["time"], ti["n_hands"])
    axes[6].plot(ti["time"], ti["n_eyes"])
    axes[7].plot(ti["time"], ti["joint_00"])
    axes[8].plot(ti["time"], ti["joint_01"])
    axes[9].plot(ti["time"], ti["joint_02"])
    axes[10].plot(ti["time"], ti["joint_29"])
    axes[11].plot(ti["time"], ti["joint_30"])

axes[0].set(title="pain survey 1", xlabel="time", ylabel="pain_survey_1")
axes[1].set(title="pain survey 2", xlabel="time", ylabel="pain_survey_2")
axes[2].set(title="pain survey 3", xlabel="time", ylabel="pain_survey_3")
axes[3].set(title="pain survey 4", xlabel="time", ylabel="pain_survey_4")
axes[4].set(title="number of legs", xlabel="time", ylabel="n_legs")
axes[5].set(title="number of hands", xlabel="time", ylabel="n_hands")
axes[6].set(title="number of eyes", xlabel="time", ylabel="n_eyes")
axes[7].set(title="joint 001", xlabel="time", ylabel="joint_00")
axes[8].set(title="joint 002", xlabel="time", ylabel="joint_01")
axes[9].set(title="joint 003 ...", xlabel="time", ylabel="joint_02")
axes[10].set(title="... joint 029", xlabel="time", ylabel="joint_29")
axes[11].set(title="joint 030", xlabel="time", ylabel="joint_30")

plt.tight_layout()
plt.show()
```

## Joint 30

```python
joint_30_train = X_train[X_train.columns[-1]].values
joint_30_test = X_test[X_test.columns[-1]].values

joint_30_train.shape, joint_30_test.shape
```

```python
count_05 = np.count_nonzero(joint_30_train == 0.5)
print(f"Number of occurrences of 0.5 in joint_30_train: {count_05}")
print("percentage of 0.5 in joint_30_train:", (count_05 / joint_30_train.shape[0]) * 100)

count_05 = np.count_nonzero(joint_30_test == 0.5)
print(f"Number of occurrences of 0.5 in joint_30_test: {count_05}")
print("percentage of 0.5 in joint_30_test:", (count_05 / joint_30_test.shape[0]) * 100)
```

we can drop the last column

```python
X_train = X_train.drop(columns=[X_train.columns[-1]])
X_test = X_test.drop(columns=[X_test.columns[-1]])
```

## Save the new data

```python
# save cleaned datasets to the data folder
X_train.to_csv(f'{DirectoryCleaned}/train.csv', index=False)
X_test.to_csv(f'{DirectoryCleaned}/test.csv', index=False)
y_train.to_csv(f'{DirectoryCleaned}/train_labels.csv', index=False)

print("Saved files:")
print(f"- {DirectoryCleaned}/train.csv (rows: {X_train.shape[0]}, cols: {X_train.shape[1]})")
print(f"- {DirectoryCleaned}/test.csv  (rows: {X_test.shape[0]},  cols: {X_test.shape[1]})")
print(f"- {DirectoryCleaned}/train_labels.csv (rows: {y_train.shape[0]}, cols: {y_train.shape[1]})")
```

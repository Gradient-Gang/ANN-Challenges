```python
import os
import re

path = "lightning_logs"

best_5_scores = []

# Find all .ckpt files in the checkpoints folders
for folder in os.listdir(path):
    in_path = os.path.join(path, folder, "checkpoints")

    # Skip if 'checkpoints' folder doesn't exist
    if not os.path.isdir(in_path):
        continue

    for file in os.listdir(in_path):
        if file.endswith(".ckpt"):
            # Improved regex: captures only digits and dots, stops before .ckpt
            match = re.search(r"val_F1=([0-9]+(?:\.[0-9]+)?)", file)
            if match:
                score = float(match.group(1))
                #print(f"Folder: {folder}, File: {file}, Score: {score}")

                if len(best_5_scores) < 5:
                    best_5_scores.append((score, os.path.join(in_path, file)))
                    best_5_scores.sort(reverse=True, key=lambda x: x[0])

                for i, (s, f) in enumerate(best_5_scores):
                    if score > s:
                        best_5_scores.insert(i, (score, os.path.join(in_path, file)))
                        break

print("Top 5 Checkpoints:")
for score, filepath in best_5_scores[:5]:
    print(f"Score: {score}, File: {filepath}")
```

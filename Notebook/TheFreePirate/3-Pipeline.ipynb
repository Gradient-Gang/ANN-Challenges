```python
from GradientGang.TheFreePirate.Piepeline.PiratePipeline import PiratePipeline
%load_ext autoreload
%autoreload 2
```

```python
dataParams = {
    'folderPath': '../../dataset/PirateProcessed',
    'trainTimeSeriesFileName': 'pirate_pain_train.csv',
    'trainGlobalFeaturesFileName': 'train_global_features.csv',
    'trainLabelsFileName': 'pirate_pain_train_labels.csv',
    'testTimeSeriesFileName': 'pirate_pain_test.csv',
    'testGlobalFeaturesFileName': 'test_global_features.csv',
    'labelsMapping': {'no_pain': 0, 'low_pain': 1, 'high_pain': 2},
    'batch_size': 64,
    'num_workers': 0,
    'columnsToIgnore': ['joint_10'],
}

callbacksParams = {
    'baseLogDir': 'PirateLogs',
    'earlyStoppingPatience': 50,
    'maxEpochs': 300,
}

pipeline = PiratePipeline("PiratePipeline_TestNotebook", dataParams, callbacksParams, seed = 42)
```

```python
phi = {
    'dataParams': {
        'numFolds': 5,
        'includeTestInFolds': True,
        'augmentTestSet': False,
        'dataAugmentationParams':{
            'nCopies': 3,
            'keepOriginal': True,
            'scaleRange': 0.06406,
            'jitterStdDev': 0.004833,
            'offsetRange': 0.01932,
            'maxWarpFraction': 0.027549,
            'windowSize': 10,
            'windowStride': 8,
        }
    },
    'modelParams': {
        'f1AverageStrategy': 'weighted',
        'numClasses': 3,
        'reconstructionLossWeight': 0.5,
        'useGlobalFeatures': False,
        'learningRate': 1e-3,
        'weightDecay': 3.4365758e-6,
        'activationFunction': 'relu',
        'timeSeriesArchitectureType': 'feedForward',
        'timeSeriesInceptionNumLayers': 2,
        'timeSeriesInceptionBaseChannels': 16,
        'timeSeriesInceptionKernelSizes': [3, 5, 7],

        'timeSeriesEncoderNumLayers': 2,
        'timeSeriesEmbeddingDim': 40,
        'timeSeriesDropout': 0.0435673,

        'predictorNumLayers': 1,
        'predictorDropout': 0.238347,
        'predictorFixALogit': True,
        'predictorLogitToFix': 0,
                
        'aggregationStrategyForClassification': 'majorityVoting',
    }
}

pipeline.setupKfoldEvaluation(phi)
```

```python
pipeline.kFoldEvaluation(phi)
```

```python
predictions = pipeline.predictFromEnsembledModel()
```
